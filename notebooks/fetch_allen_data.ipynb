```python
import abagen
import pandas as pd
```

```python
atlas = abagen.fetch_desikan_killiany()
atlas_image = atlas['image']
atlas_info = pd.read_csv(atlas['info'])
atlas_info.to_csv("../datasets/processed/dk_atlas_info.csv", index=False)
print(atlas_info.columns)
atlas_info.head()
```

```python
donor_ids = ['9861', '10021', '12876', '14380', '15496', '15697']

for donor in donor_ids:
    try:
        abagen.fetch_microarray(donors=[donor], resume=True)
        print(f"Donor {donor} done")
    except Exception as e:
        print(f"Donor {donor} failed: {e}")
```

```python
expression = abagen.get_expression_data(atlas_image, atlas_info)
print(expression.shape)
expression.head()
```

```python
expression.to_csv("../datasets/processed/parkinsons_expression_dk.csv")
```
